```python
from pathlib import Path

sql_file = Path("supermarket_data_clean.sql")

with open(sql_file, "w", encoding="utf-8") as f:

    for _, row in sql_df.iterrows():

        values = []

        for column in sql_df.columns:

            value = row[column]

            if pd.isna(value):
                values.append("NULL")

            elif column in ["order_date", "ship_date"]:
                values.append(f"'{value.strftime('%Y-%m-%d')}'")

            elif column in ["sales", "profit"]:
                values.append(f"{float(value):.4f}")

            elif isinstance(value, str):
                value = value.replace("'", "''")
                values.append(f"'{value}'")

            else:
                values.append(str(value))

        f.write(
            "INSERT INTO staging_sales VALUES ("
            + ", ".join(values)
            + ");\n"
        )

print("Clean SQL file created successfully!")
print("Location:", sql_file.resolve())
```
